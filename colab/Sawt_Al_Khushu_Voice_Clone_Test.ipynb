```python
# رفع التسجيل من الهاتف إلى جلسة Colab الخاصة
from google.colab import files
uploaded = files.upload()
audio_path = next(iter(uploaded.keys()))
print('تم اختيار الملف:', audio_path)
```

```python
# تحميل النموذج وتوليد عينة بصوتك
import sys, os, getpass
sys.path.insert(0, '/content/Audar-TTS-V1')
import soundfile as sf
from audar_tts import load_model, load_codec, synthesize, OUT_SR

reference_text = input('اكتب الكلمات التي نطقتها حرفياً في التسجيل: ').strip()
target_text = input('اكتب النص الذي تريد سماعه بصوتك: ').strip()
if not reference_text or not target_text:
    raise ValueError('يجب إدخال نص التسجيل والنص المراد قراءته.')

print('تحميل نموذج Audar Flash — قد يستغرق عدة دقائق...')
model, tok = load_model('audarai/Audar-TTS-V1-Flash')
codec = load_codec()
print('جارٍ توليد الصوت...')
wav = synthesize(model, tok, codec, '[whispers] ' + target_text,
                 ref_wav=audio_path, ref_text=reference_text,
                 temperature=0.65)
output_path = '/content/sawt_al_khushu_test.wav'
sf.write(output_path, wav, OUT_SR)
print('اكتمل إنشاء العينة:', output_path)
```

```python
# الاستماع والتنزيل إلى الهاتف
from IPython.display import Audio, display
display(Audio('/content/sawt_al_khushu_test.wav'))
files.download('/content/sawt_al_khushu_test.wav')
```
