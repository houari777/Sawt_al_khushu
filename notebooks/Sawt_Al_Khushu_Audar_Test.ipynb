```python
import torch
print('GPU available:', torch.cuda.is_available())
if not torch.cuda.is_available():
    raise RuntimeError('فعّل GPU من Runtime > Change runtime type > T4 GPU ثم أعد تشغيل هذه الخلية.')
```

```python
from google.colab import files
uploaded = files.upload()
if not uploaded:
    raise RuntimeError('ارفع ملف demo_male_2_ar.wav الذي نزّلته من عينة Audar.')
ref_file = next(iter(uploaded))
print('ملف الصوت المرجعي:', ref_file)
```

```python
import torch
import soundfile as sf
from audar_tts import load_model, load_codec, synthesize, OUT_SR

# T4 يحتاج float16؛ نستخدم GPU المتاح في Colab
device = 'cuda:0'
model, tok = load_model('audarai/Audar-TTS-V1-Flash', device=device, dtype=torch.float16)
codec = load_codec(device=device)

# النص المطابق للعينة العربية demo_male_2 التي رفعتها
ref_text = 'تعال، اقترب قليلاً لقد خطّطتُ لمفاجأةٍ رائعة طوال الأسبوع، وأنا واثقٌ تماماً أنّها ستُدهشك حقاً!'
target_text = input('اكتب جملة عربية قصيرة لاختبار القراءة السرية: ').strip()
if not target_text:
    target_text = 'بِسْمِ اللَّهِ الرَّحْمَٰنِ الرَّحِيمِ. الْحَمْدُ لِلَّهِ رَبِّ الْعَالَمِينَ.'

wav = synthesize(model, tok, codec, '[whispers] ' + target_text,
                 ref_wav=ref_file, ref_text=ref_text, temperature=0.65)
sf.write('sawt_al_khushu_test.wav', wav, OUT_SR)
print('تم إنشاء الصوت التجريبي')
```
